```python
class Robot :
  def __init__(self,name):
    self.name = name
  def sound(self) :
    print("Sound")

marvin = Robot("Marvin")
```

```python
class MinigRobot(Robot):
  def __init__(self, name, tool):
    super().__init__(name)
    self.tool = tool
  def sound (self):
    print("Drill sound")
```

```python
# Here is a simple, highly engaging coding task tailored
# perfectly for your students. It directly follows our RPG Game Engine
# story, giving them a practical win without getting
# bogged down in complex logic.
# 🏆 The Task: Build a Pokemon-Style Battle Monster!
# 📝 The Scenario
# You are building a text-based monster battle game.
# You need to create a base monster class, inherit from it to
# create a specific element type, and protect its health pool so
# players can't cheat.
# 💻 Step-by-Step Instructions for Students
# Step 1: The Base Monster (class)
# Create a class named Monster.
# It should take name during initialization.
# It should have a method called speak(self) that prints:
# "[Monster Name] makes a generic monster noise!"
# Step 2: The Fire Dragon (Inheritance)
# Create a child class named FireMonster that inherits from Monster.
# Override the speak(self) method so a fire monster says:
#  "[Monster Name] breathes a blast of fire! 🔥"
```

```python
class Monster :
  def __init__(self, name) -> None:
    self.name = name
    self.__health = 100

  def speak (self,sound) :
    print(f"{self.name} makes a generic monster noise!")

  @property
  def health(self):
    print(self.__health)

  @health.setter
  def health(self, amount) :
    print("I am in setter")
    self.__health = amount


  def take_damage(self, damage_point) :
    self.__health = self.__health - damage_point
```

```python
class Fire_Dragon (Monster) :
  def speak(self) :
    print(f"{self.name} breathes a blast of fire! 🔥")
```

```python
f = Fire_Dragon("fire_dragon")
f.speak()
```

```python
f.take_damage(10)
f.health
```
