```python
scores = {
    "Alice": 85,
    "Bob": 62,
    "Charlie": 95,
    "David": 48,
    "Eva": 73
}

for student in scores:
    mark = scores[student]

    if mark >= 90:
        result = "A"
    elif mark >= 80:
        result = "B"
    elif mark >= 70:
        result = "C"
    elif mark >= 60:
        result = "D"
    else:
        result = "F"

    print(student, "received a grade of", result)
```

```python
inventory = {
    "laptop": 15,
    "mouse": 0,
    "keyboard": 4,
    "monitor": 12,
    "hdmi_cable": 8
}

for product in inventory:
    quantity = inventory[product]

    if quantity == 0:
        print(product, ": ALERT - Out of Stock!")
    elif quantity < 10:
        print(product, ": ALERT - Urgent Reorder Required (", quantity, "units)")
    else:
        print(product, ": Sufficiently Stocked (", quantity, "units)")
```

```python
prices = {
    "Coffee": 4.50,
    "Book": 12.00,
    "Headphones": 45.00,
    "Sticker": 1.50,
    "Smartwatch": 120.00
}

for item in prices:
    cost = prices[item]

    if cost > 20:
        cost = cost * 1.1
    elif cost < 10:
        cost = cost * 0.95

    prices[item] = round(cost, 2)

for item in list(prices):
    if prices[item] == 0:
        del prices[item]

print(prices)
```
