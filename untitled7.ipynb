```python
%%writefile app.py
import streamlit as st

class BankAccount:

    def __init__(self, account_holder, balance=0.0):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            return True
        return False

    def display_balance(self):
        return f"Account holder: {self.account_holder} | Current Balance: ${self.balance}"

if "database" not in st.session_state:
    st.session_state.database = {}

st.title("Simple Bank Account")

st.subheader("Create New User")

user_name = st.text_input("Enter User Name")
user_balance = st.number_input("Enter Balance", min_value=0.0)

if st.button("Register"):
    if user_name == "":
        st.error("Enter user name")
    elif user_name in st.session_state.database:
        st.error("User already exists")
    else:
        st.session_state.database[user_name] = BankAccount(user_name, user_balance)
        st.success("Account Created")

if len(st.session_state.database) > 0:

    user = st.selectbox("Choose User", list(st.session_state.database.keys()))

    amount = st.number_input("Amount", min_value=0.0)

    if st.button("Deposit"):
        st.session_state.database[user].deposit(amount)
        st.success("Deposit Successful")

    if st.button("Withdraw"):
        if st.session_state.database[user].withdraw(amount):
            st.success("Withdraw Successful")
        else:
            st.error("Insufficient funds")

    if st.button("Display Balance"):
        st.write(st.session_state.database[user].display_balance())
```
