```python
# Install Gradio
!pip -q install gradio==5.38.2

import gradio as gr

students = []


class Student:
    def init(self, name, age, student_id):
        self.name = name
        self.age = int(age)
        self.student_id = student_id
        self.grades = []

    def add_grade(self, grade):
        self.grades.append(float(grade))

    def calculate_average(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)

    def get_status(self):
        avg = self.calculate_average()

        if avg >= 90:
            return "Excellent"
        elif avg >= 80:
            return "Very Good"
        elif avg >= 70:
            return "Good"
        elif avg >= 60:
            return "Pass"
        else:
            return "Fail"

    def get_gpa(self):
        avg = self.calculate_average()

        if avg >= 90:
            return 4.0
        elif avg >= 80:
            return 3.5
        elif avg >= 70:
            return 3.0
        elif avg >= 60:
            return 2.0
        else:
            return 0.0

    def display_info(self):
        return f"""
Name: {self.name}
Age: {self.age}
Student ID: {self.student_id}
Grades: {self.grades}
Average: {self.calculate_average():.2f}
Status: {self.get_status()}
GPA: {self.get_gpa()}
"""


def add_student(name, age, student_id):
    if name == "" or student_id == "":
        return "Please fill all fields."

    for s in students:
        if s.student_id == student_id:
            return "Student ID already exists."

    students.append(Student(name, age, student_id))
    return "Student added successfully."


def add_grade(student_id, grade):
    for s in students:
        if s.student_id == student_id:
            s.add_grade(grade)
            return "Grade added successfully."

    return "Student not found."


def show_students():
    if len(students) == 0:
        return "No students."

    result = ""

    for s in students:
        result += s.display_info()
        result += "\n------------------------\n"

    return result

def search_student(student_id):
    for s in students:
        if s.student_id == student_id:
            return s.display_info()

    return "Student not found."


def delete_student(student_id):
    for s in students:
        if s.student_id == student_id:
            students.remove(s)
            return "Student deleted successfully."

    return "Student not found."


def highest_average():
    if len(students) == 0:
        return "No students."

    best = students[0]

    for s in students:
        if s.calculate_average() > best.calculate_average():
            best = s

    return f"""
Highest Average Student

Name: {best.name}
Student ID: {best.student_id}
Average: {best.calculate_average():.2f}
Status: {best.get_status()}
GPA: {best.get_gpa()}
"""


def count_students():
    passed = 0
    failed = 0

    for s in students:
        if s.calculate_average() >= 60:
            passed += 1
        else:
            failed += 1

    return f"""
Passed Students : {passed}
Failed Students : {failed}
"""


def sort_students():
    if len(students) == 0:
        return "No students."

    sorted_list = sorted(
        students,
        key=lambda x: x.calculate_average(),
        reverse=True
    )

    result = ""

    for s in sorted_list:
        result += f"{s.name} - {s.calculate_average():.2f}\n"

    return result

with gr.Blocks(title="Student Management System") as demo:

    gr.Markdown("# 🎓 Student Management System")

    with gr.Tab("Add Student"):
        name = gr.Textbox(label="Student Name")
        age = gr.Number(label="Age")
        student_id = gr.Textbox(label="Student ID")

        add_btn = gr.Button("Add Student")
        add_output = gr.Textbox(label="Result")

        add_btn.click(
            fn=add_student,
            inputs=[name, age, student_id],
            outputs=add_output
        )

    with gr.Tab("Add Grade"):
        grade_id = gr.Textbox(label="Student ID")
        grade = gr.Number(label="Grade")

        grade_btn = gr.Button("Add Grade")
        grade_output = gr.Textbox(label="Result")

        grade_btn.click(
            fn=add_grade,
            inputs=[grade_id, grade],
            outputs=grade_output
        )

    with gr.Tab("Show Students"):
        show_btn = gr.Button("Show All Students")
        show_output = gr.Textbox(lines=18)

        show_btn.click(
            fn=show_students,
            outputs=show_output
        )

    with gr.Tab("Search Student"):
        search_id = gr.Textbox(label="Student ID")

        search_btn = gr.Button("Search")
        search_output = gr.Textbox(lines=10)

        search_btn.click(
            fn=search_student,
            inputs=search_id,
            outputs=search_output
        )

    with gr.Tab("Delete Student"):
        delete_id = gr.Textbox(label="Student ID")

        delete_btn = gr.Button("Delete Student")
        delete_output = gr.Textbox()

        delete_btn.click(
            fn=delete_student,
            inputs=delete_id,
            outputs=delete_output
        )

    with gr.Tab("Highest Average"):
        highest_btn = gr.Button("Show Highest Average")
        highest_output = gr.Textbox(lines=8)

        highest_btn.click(
            fn=highest_average,
            outputs=highest_output
        )

    with gr.Tab("Passed / Failed"):
        count_btn = gr.Button("Count Students")
        count_output = gr.Textbox()

        count_btn.click(
            fn=count_students,
            outputs=count_output
        )

    with gr.Tab("Sort Students"):
        sort_btn = gr.Button("Sort Students")
        sort_output = gr.Textbox(lines=15)

        sort_btn.click(
            fn=sort_students,
            outputs=sort_output)

demo.launch(
    share=True,
    debug=True
)
```
