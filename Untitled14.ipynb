```python
!pip install gradio -q

import gradio as gr

students = []


class Student:
    def __init__(self, name, age, student_id):
        self.name = name
        self.age = age
        self.student_id = student_id
        self.grades = []

    def add_grade(self, grade):
        self.grades.append(grade)

    def calculate_average(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)

    def get_status(self):
        avg = self.calculate_average()

        if avg >= 90:
            return "Excellent"
        elif avg >= 80:
            return "Very Good"
        elif avg >= 70:
            return "Good"
        elif avg >= 60:
            return "Pass"
        else:
            return "Fail"

    def get_gpa(self):
        avg = self.calculate_average()

        if avg >= 90:
            return 4.0
        elif avg >= 80:
            return 3.5
        elif avg >= 70:
            return 3.0
        elif avg >= 60:
            return 2.0
        else:
            return 0.0


def add_student(name, age, student_id):

    if name == "" or student_id == "":
        return "Please fill all fields."

    student = Student(name, int(age), student_id)
    students.append(student)

    return "Student Added Successfully."


def add_grade(student_id, grade):

    for student in students:
        if student.student_id == student_id:
            student.add_grade(float(grade))
            return "Grade Added Successfully."

    return "Student Not Found."


def show_students():

    if len(students) == 0:
        return "No Students."

    text = ""

    for student in students:

        text += f"Name: {student.name}\n"
        text += f"Age: {student.age}\n"
        text += f"Student ID: {student.student_id}\n"
        text += f"Grades: {student.grades}\n"
        text += f"Average: {round(student.calculate_average(),2)}\n"
        text += f"Status: {student.get_status()}\n"
        text += f"GPA: {student.get_gpa()}\n"
        text += "-----------------------------\n"

    return text


def search_student(student_id):

    for student in students:

        if student.student_id == student_id:

            return f"""
Name: {student.name}
Age: {student.age}
Student ID: {student.student_id}
Grades: {student.grades}
Average: {round(student.calculate_average(),2)}
Status: {student.get_status()}
GPA: {student.get_gpa()}
"""

    return "Student Not Found."


def delete_student(student_id):

    for student in students:

        if student.student_id == student_id:
            students.remove(student)
            return "Student Deleted Successfully."

    return "Student Not Found."


def highest_average():

    if len(students) == 0:
        return "No Students."

    best = students[0]

    for student in students:

        if student.calculate_average() > best.calculate_average():
            best = student

    return f"""
Top Student

Name: {best.name}
Average: {round(best.calculate_average(),2)}
Status: {best.get_status()}
GPA: {best.get_gpa()}
"""


def count_students():

    passed = 0
    failed = 0

    for student in students:

        if student.calculate_average() >= 60:
            passed += 1
        else:
            failed += 1

    return f"Passed Students: {passed}\nFailed Students: {failed}"

def sort_students():

    if len(students) == 0:
        return "No Students."

    students.sort(key=lambda x: x.calculate_average(), reverse=True)

    text = ""

    for student in students:
        text += f"{student.name}  |  Average: {round(student.calculate_average(),2)}\n"

    return text


with gr.Blocks(title="Student Management System") as demo:

    gr.Markdown("# Student Management System")

    with gr.Tab("Add Student"):

        name = gr.Textbox(label="Student Name")
        age = gr.Number(label="Age")
        student_id = gr.Textbox(label="Student ID")

        add_button = gr.Button("Add Student")

        add_result = gr.Textbox(label="Result")

        add_button.click(
            fn=add_student,
            inputs=[name, age, student_id],
            outputs=add_result
        )

    with gr.Tab("Add Grade"):

        grade_id = gr.Textbox(label="Student ID")
        grade = gr.Number(label="Grade")

        grade_button = gr.Button("Add Grade")

        grade_result = gr.Textbox(label="Result")

        grade_button.click(
            fn=add_grade,
            inputs=[grade_id, grade],
            outputs=grade_result
        )

    with gr.Tab("Show Students"):

        show_button = gr.Button("Show All Students")

        show_result = gr.Textbox(lines=20)

        show_button.click(
            fn=show_students,
            outputs=show_result
        )

    with gr.Tab("Search Student"):

        search_id = gr.Textbox(label="Student ID")

        search_button = gr.Button("Search")

        search_result = gr.Textbox(lines=10)

        search_button.click(
            fn=search_student,
            inputs=search_id,
            outputs=search_result
        )

    with gr.Tab("Delete Student"):

        delete_id = gr.Textbox(label="Student ID")

        delete_button = gr.Button("Delete Student")

        delete_result = gr.Textbox()

        delete_button.click(
            fn=delete_student,
            inputs=delete_id,
            outputs=delete_result
        )

    with gr.Tab("Highest Average"):

        highest_button = gr.Button("Show Highest Average")

        highest_result = gr.Textbox(lines=8)

        highest_button.click(
            fn=highest_average,
            outputs=highest_result
        )

    with gr.Tab("Passed / Failed"):

        count_button = gr.Button("Count Students")

        count_result = gr.Textbox()

        count_button.click(
            fn=count_students,
            outputs=count_result
        )

    with gr.Tab("Sort Students"):

        sort_button = gr.Button("Sort")

        sort_result = gr.Textbox(lines=15)

        sort_button.click(
            fn=sort_students,
            outputs=sort_result)


demo.launch()
```
