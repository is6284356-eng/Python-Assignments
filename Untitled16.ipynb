```python
import pandas as pd

# =========================
# Task 1
# =========================

numbers = pd.Series([10, 20, 30, 40, 50])

print("Values:")
print(numbers.values)

print("\nIndex:")
print(numbers.index)

print("\nData Type:")
print(numbers.dtype)


# =========================
# Task 2
# =========================

marks = pd.Series(
    [90, 95, 88, 92],
    index=["Ahmed", "Sara", "Ali", "Mona"]
)

print("\nSara's Mark:")
print(marks["Sara"])


# =========================
# Task 3
# =========================

students = {
    "Name": ["Ahmed", "Sara", "Ali", "Mona", "Omar"],
    "Age": [20, 21, 19, 22, 20],
    "City": ["Cairo", "Alex", "Giza", "Tanta", "Mansoura"]
}

df = pd.DataFrame(students)

print("\nDataFrame:")
print(df)

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```
