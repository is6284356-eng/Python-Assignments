```python
# user choice operation to do
num1 = int(input("Enter num1"))
num2 = int(input("Enter num2"))
operand = input("Enter operand")
if operand == '+' :
  print(num1+num2)
elif operand == '-' :
  print(num1 - num2)
elif operand == '/' :
  print(num1/ num2)
elif operand == '*' :
  print(num1 * num2)
else :
  print("Unsupported operand")
```

```python
# take  3 input from use  name , age , salary ---> save it to database
# show all user info in database
database = {}
n = int(input("Enter number of users"))
for i in range(n) :
  name = input("Enter your name")
  age  = int(input("Enter your Age"))
  salary = float(input("Enter your salary"))
  database[name] = {"salary":salary, 'Age':age}
```
