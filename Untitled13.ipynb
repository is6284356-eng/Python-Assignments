```python
class Student:
    def __init__(self, name, age, student_id):
        self.name = name
        self.age = age
        self.student_id = student_id
        self.grades = []

    def add_grade(self, score):
        self.grades.append(score)

    def calculate_average(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)

    def get_status(self):
        average = self.calculate_average()

        if average >= 90:
            return "Excellent"
        elif average >= 80:
            return "Very Good"
        elif average >= 70:
            return "Good"
        elif average >= 60:
            return "Pass"
        else:
            return "Fail"

    def get_gpa(self):
        average = self.calculate_average()

        if average >= 90:
            return 4.0
        elif average >= 80:
            return 3.5
        elif average >= 70:
            return 3.0
        elif average >= 60:
            return 2.0
        else:
            return 0.0

    def display_info(self):
        print("\nStudent Information")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Student ID:", self.student_id)
        print("Grades:", self.grades)
        print("Average:", round(self.calculate_average(), 2))
        print("Status:", self.get_status())
        print("GPA:", self.get_gpa())
        print("-" * 30)


def create_student():
    name = input("Enter student name: ")
    age = int(input("Enter age: "))
    student_id = input("Enter student ID: ")

    return Student(name, age, student_id)


def add_grades(student):
    number = int(input("How many grades? "))

    for i in range(number):
        grade = float(input(f"Enter grade {i + 1}: "))
        student.add_grade(grade)


def show_all_students(student_list):
    if len(student_list) == 0:
        print("No students found.")
    else:
        for student in student_list:
            student.display_info()


def search_student(student_list):
    search_id = input("Enter Student ID: ")

    found = False

    for student in student_list:
        if student.student_id == search_id:
            student.display_info()
            found = True
            break

    if not found:
        print("Student not found.")


def delete_student(student_list):
    student_id = input("Enter Student ID to delete: ")

    for student in student_list:
        if student.student_id == student_id:
            student_list.remove(student)
            print("Student deleted successfully.")
            return

    print("Student not found.")


def highest_average(student_list):
    if len(student_list) == 0:
        print("No students found.")
        return

    top_student = student_list[0]

    for student in student_list:
        if student.calculate_average() > top_student.calculate_average():
            top_student = student

    print("\nTop Student")
    top_student.display_info()


def count_students(student_list):
    passed = 0
    failed = 0

    for student in student_list:
        if student.calculate_average() >= 60:
            passed += 1
        else:
            failed += 1

    print("Passed Students:", passed)
    print("Failed Students:", failed)


def sort_students(student_list):
    student_list.sort(key=lambda student: student.calculate_average(), reverse=True)

    print("Students sorted by average.")

    for student in student_list:
        student.display_info()


students = []

while True:
    print("\n===== Student Management System =====")
    print("1. Add Student")
    print("2. Add Grades")
    print("3. Show All Students")
    print("4. Search Student")
    print("5. Delete Student")
    print("6. Highest Average")
    print("7. Count Passed/Failed Students")
    print("8. Sort Students by Average")
    print("9. Exit")

    choice = input("Choose: ")

    if choice == "1":
        student = create_student()
        students.append(student)
        print("Student added successfully.")

    elif choice == "2":
        sid = input("Enter Student ID: ")

        found = False

        for student in students:
            if student.student_id == sid:
                add_grades(student)
                found = True
                break

        if not found:
            print("Student not found.")

    elif choice == "3":
        show_all_students(students)

    elif choice == "4":
        search_student(students)

    elif choice == "5":
        delete_student(students)

    elif choice == "6":
        highest_average(students)

    elif choice == "7":
        count_students(students)

    elif choice == "8":
        sort_students(students)

    elif choice == "9":
        print("Program closed.")
        break

    else:
        print("Invalid choice.")
```
